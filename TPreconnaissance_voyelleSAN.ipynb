FONCTION QUI RETOURNE L'AMPLITUDE ET LA FREQUENCE DU FONDAMENTALE ET DE H HARMONIQUES SOUS FORME DE VECTEUR

```python
signal_dict = np.load('a_1.npz')

s = signal_dict['s']
t = signal_dict['t']


df=pd.DataFrame({"t":t, "s":s})
fig= px.line(df, x="t", y="s", title="elgazouz")
fig.show()

nu, S = mana.trans_fourier(s,t)


spectrenormalS = ((np.abs(S))**2)/((np.abs(np.max(S)))**2)

df=pd.DataFrame({"t":t, "spectrenormalS":spectrenormalS})
fig2= px.line(df, x="t", y="spectrenormalS", title="elgazouz")
fig2.show()

spectrelogS = np.log(spectrenormalS)

df=pd.DataFrame({"t":t, "spectrelogS":spectrelogS})
fig3= px.line(df, x="t", y="spectrelogS", title="elgazouz")
fig3.show()
#prend les pics
```
