MOYENNE AMPLITUDE LETTRE A

```python


L = [
    'a_1.npz',
    'apprentissage/a_2.npz',
    'apprentissage/a_3.npz',
    'apprentissage/a_4.npz',
    'apprentissage/a_5.npz'
]

def moyenneliste(m, h=10):
    toutes_les_amplitudes = []  #

    for fichier in m:
        
        signal_dict = np.load(fichier)
        s = signal_dict['s']
        t = signal_dict['t']

        
        nu, S = san.trans_fourier(s, t)
        spectreS = (np.abs(S))**2
        SpectreSN = spectreS / np.max(spectreS)
        spectrelogS = np.log10(spectreS) / np.max(np.log10(spectreS))

        
        f0 = san.detect_fondamentale(SpectreSN, nu)
        liste_freq = [k * f0 for k in range(1, h + 1)]
        
        freq, amp = san.detect_pics(spectrelogS, liste_freq, nu)
        
        
        toutes_les_amplitudes.append(amp)
        print(f"Amplitudes pour {fichier} :", amp)

    
    moyenne_harmoniques = np.mean(toutes_les_amplitudes, axis=0)
    
    return moyenne_harmoniques


moyenneA = moyenneliste(L, h=10)
print("Moyenne des amplitudes par harmonique :", moyenneA)
A=0
for i in range(len(moyenneA)):
    A+=moyenneA[i]
Afinal=A/len(moyenneA)
print("Moyenne de A :", Afinal)
```

MOYENNE AMPLITUDE LETTRE O

```python



L = [
    'apprentissage/o_1.npz',
    'apprentissage/o_2.npz',
    'apprentissage/o_3.npz',
    'apprentissage/o_4.npz',
    'apprentissage/o_5.npz'
]

def moyenneliste(m, h=10):
    toutes_les_amplitudes = []  # Va contenir la liste des amplitudes de chaque fichier

    for fichier in m:
        # 1. Chargement du signal courant (variables locales écrasées à chaque tour)
        signal_dict = np.load(fichier)
        s = signal_dict['s']
        t = signal_dict['t']

        # 2. Calculs spectraux
        nu, S = san.trans_fourier(s, t)
        spectreS = (np.abs(S))**2
        SpectreSN = spectreS / np.max(spectreS)
        spectrelogS = np.log10(spectreS) / np.max(np.log10(spectreS))

        # 3. Détection de f0 et des h harmoniques
        f0 = san.detect_fondamentale(SpectreSN, nu)
        liste_freq = [k * f0 for k in range(1, h + 1)]
        
        freq, amp = san.detect_pics(spectrelogS, liste_freq, nu)
        
        # On ajoute la liste des amplitudes de ce fichier à notre liste globale
        toutes_les_amplitudes.append(amp)
        print(f"Amplitudes pour {fichier} :", amp)

    # 4. Calcul de la moyenne pour chaque harmonique (axis=0 fait la moyenne colonne par colonne)
    moyenne_harmoniques = np.mean(toutes_les_amplitudes, axis=0)
    
    return moyenne_harmoniques

# Exécution de la fonction
moyenneO = moyenneliste(L, h=10)
print("Moyenne des amplitudes par harmonique :", moyenneO)
O=0
for i in range(len(moyenneO)):
    O+=moyenneO[i]
Ofinal=O/len(moyenneO)
print("Moyenne de O :", Ofinal)
```
